```python
print("Missing cities after cleaning:", matches["city"].isnull().sum())
```

```python
# Convert date column into datetime format
matches["date"] = pd.to_datetime(matches["date"])

# Check the data type
print(matches["date"].dtype)
```

```python

# Extract year and month from date
matches["year"] = matches["date"].dt.year
matches["month"] = matches["date"].dt.month

# Display selected columns
print(matches[["date", "year", "month"]].head())
```

```python
# Get unique team names
teams = pd.concat([
    matches["team1"],
    matches["team2"]
]).unique()

print("Total Unique Team Names:", len(teams))

for team in sorted(teams):
    print(team)
```

```python
# Dictionary for team name standardization

team_mapping = {
    "Delhi Daredevils": "Delhi Capitals",
    "Kings XI Punjab": "Punjab Kings",
    "Royal Challengers Bangalore": "Royal Challengers Bengaluru",
    "Rising Pune Supergiant": "Rising Pune Supergiants"
}

# Replace names in matches dataset
matches["team1"] = matches["team1"].replace(team_mapping)
matches["team2"] = matches["team2"].replace(team_mapping)
matches["toss_winner"] = matches["toss_winner"].replace(team_mapping)
matches["winner"] = matches["winner"].replace(team_mapping)

# Check updated team names
teams = pd.concat([
    matches["team1"],
    matches["team2"]
]).unique()

print("Total Unique Team Names:", len(teams))

for team in sorted(teams):
    print(team)
```

```python
deliveries["batting_team"] = deliveries["batting_team"].replace(team_mapping)
deliveries["bowling_team"] = deliveries["bowling_team"].replace(team_mapping)
```

```python
print("MATCHES DATASET")
print("Team 1:", matches["team1"].nunique())
print("Team 2:", matches["team2"].nunique())

print("\nDELIVERIES DATASET")
print("Batting Teams:", deliveries["batting_team"].nunique())
print("Bowling Teams:", deliveries["bowling_team"].nunique())
```

```python
# Count wins for each team
team_wins = matches["winner"].value_counts()

# Display top 10 teams
print(team_wins.head(10))
```

```python
import matplotlib.pyplot as plt

# Get top 10 teams
top_teams = team_wins.head(10)

# Create bar chart
plt.figure(figsize=(12, 6))

plt.barh(top_teams.index, top_teams.values)

# Add labels and title
plt.xlabel("Number of Wins")
plt.ylabel("IPL Teams")
plt.title("Top 10 IPL Teams by Total Match Wins")

# Show highest wins at the top
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```
